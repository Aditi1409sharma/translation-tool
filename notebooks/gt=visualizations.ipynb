```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load results
df = pd.read_csv("../outputs_steps/batch_results_with_gt.csv")

# Sort by F1
df_sorted = df.sort_values("F1")

# ------------------------ F1 Score Line Plot ------------------------
plt.figure(figsize=(12, 5))
sns.lineplot(data=df_sorted, x=range(len(df_sorted)), y="F1")
plt.title("F1 Score per Image")
plt.xlabel("Image Index (sorted by F1)")
plt.ylabel("F1 Score")
plt.grid(True)
plt.tight_layout()
plt.show()

# ------------------------ Precision vs Recall ------------------------
plt.figure(figsize=(6, 6))
sns.scatterplot(data=df, x="Recall", y="Precision", hue="F1", palette="coolwarm", s=60)
plt.title("Precision vs Recall")
plt.grid(True)
plt.tight_layout()
plt.show()

# ------------------------ Histogram of Confidence ------------------------
plt.figure(figsize=(8, 5))
sns.histplot(df["Average_Confidence"], bins=20, kde=True)
plt.title("Distribution of Average Confidence Scores")
plt.xlabel("Average Confidence")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# ------------------------ Top and Bottom 5 Images by F1 Score ------------------------
print("Top 5 images by F1 score:")
print(df_sorted.tail(5)[["Image", "F1"]])

print("\nBottom 5 images by F1 score:")
print(df_sorted.head(5)[["Image", "F1"]])
```
