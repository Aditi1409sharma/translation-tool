```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/aditi/translation tool/output_steps/batch_results.csv")
```

```python
# Summary Stats
print("Total Images:", len(df))
print("Mean Total Boxes:", df["Total_Boxes"].mean())
print("Mean Retained Boxes:", df["Retained_Boxes"].mean())
print("Mean Confidence:", df["Average_Confidence"].mean())
print("Retention Rate (mean):", (df["Retained_Boxes"] / df["Total_Boxes"]).mean())
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Average_Confidence"], bins=20, color='skyblue', edgecolor='black')
plt.title("Distribution of Average Confidence Scores")
plt.xlabel("Average Confidence")
plt.ylabel("Number of Images")
plt.grid(True)
plt.show()
```

```python
df["Retention_Rate"] = df["Retained_Boxes"] / df["Total_Boxes"]

plt.figure(figsize=(8, 5))
plt.hist(df["Retention_Rate"], bins=20, color='lightgreen', edgecolor='black')
plt.title("Retention Rate Distribution")
plt.xlabel("Retention Rate")
plt.ylabel("Number of Images")
plt.grid(True)
plt.show()
```

```python
low_conf_df = df[df["Average_Confidence"] < 0.5]
print(f"\nImages with low average confidence (<0.5): {len(low_conf_df)}")
print(low_conf_df[["Image", "Average_Confidence"]].head())
```

```python
zero_retained_df = df[df["Retained_Boxes"] == 0]
print(f"\nImages with zero retained boxes: {len(zero_retained_df)}")
print(zero_retained_df[["Image", "Total_Boxes"]].head())
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Total_Boxes"], df["Average_Confidence"], alpha=0.7)
plt.title("Confidence vs. Total Detected Boxes")
plt.xlabel("Total Boxes")
plt.ylabel("Average Confidence")
plt.grid(True)
plt.show()
```

```python
poor_cases = df[df["Average_Confidence"] < 0.5]["Image"].tolist()
with open("C:/Users/aditi/translation tool/outputs_steps/fine_tunedy_low_confidence_images.txt", "w") as f:
    for name in poor_cases:
        f.write(name + "\n")
        print(name)
```
